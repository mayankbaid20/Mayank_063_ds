```python
# Task 1: Arrays, Statistics, and Indexing

import numpy as np

# 1. Random seed & Array A (100 integers)
np.random.seed(42)
array_a = np.random.randint(1, 101, size=100)

# 2. Statistics for Array A
print("Min    :", np.min(array_a))
print("Max    :", np.max(array_a))
print("Median :", np.median(array_a))
print("Mean   :", np.mean(array_a))
print("Std Dev:", np.std(array_a))

# 3. Array B with np.arange()
array_b = np.arange(100)

# 4. Zeros & Ones
z = np.zeros((3, 4))
o = np.ones((2, 5))
print("Zeros -> Shape:", z.shape, "| Dtype:", z.dtype)
print("Ones  -> Shape:", o.shape, "| Dtype:", o.dtype)

# 5. Linspace vs Arange
lin = np.linspace(0, 10, 5)
print("Linspace:", lin)
# Note: np.arange uses step size; np.linspace uses total number of points.

# 6. 2D & 3D Arrays + Slicing
arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
arr_3d = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])

print("2D -> Shape:", arr_2d.shape, "| ndim:", arr_2d.ndim)
print("2D Row 0, Cols 1..2:", arr_2d[0, 1:])

print("3D -> Shape:", arr_3d.shape, "| ndim:", arr_3d.ndim)

# 7. Reshape to Matrix & Flatten back
matrix = array_b.reshape(10, 10)
flat = matrix.flatten()
print("Reshaped Matrix Shape:", matrix.shape)
print("Flattened 1D Shape   :", flat.shape)
```

task 2

```python

import numpy as np

# 1. Create two square matrices
a = np.array([[4, 7], [2, 6]])
b = np.array([[0, 3], [1, 5]])

# 2. Addition, Element-wise Multiplication, Matrix Multiplication
print("Addition:\n", a + b)
print("\nElement-wise Multiplication:\n", a * b)
print("\nMatrix Multiplication (@):\n", a @ b)

# 3. Transpose & Determinant
trans = a.T
det = np.linalg.det(a)
print("\nTranspose of A:\n", trans)
print("Determinant of A:", round(det, 2))

# 4. Inverse & Verification (A @ A^-1 = Identity)
if det != 0:
    inv = np.linalg.inv(a)
    print("\nInverse of A:\n", inv)
    
    # Check if A @ A^-1 is equal to Identity Matrix
    is_identity = np.allclose(a @ inv, np.eye(2))
    print("A @ A^-1 equals Identity Matrix?:", is_identity)
```

```python
# Task 3: Normal Distribution and Histogram

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Chosen Mean & Standard Deviation
mean_val = 50.0
std_val = 15.0

# Generate 1,000 values from normal distribution
np.random.seed(42)
samples = np.random.normal(loc=mean_val, scale=std_val, size=1000)

# 2. Compute Sample Mean & Sample Std
s_mean = np.mean(samples)
s_std = np.std(samples)

print("Chosen Mean :", mean_val, "| Sample Mean :", (s_mean))
print("Chosen Std  :", std_val,  "| Sample Std  :", (s_std))

# 3. Plot Histogram
plt.figure(figsize=(8, 4))
sns.histplot(samples, kde=True, color="teal")
plt.title("Task 3: Normal Distribution (Mean=50, Std=15)")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.savefig("normal_distribution.png", bbox_inches="tight")
plt.show()
```

task 7

```python
# Task 7: Missing Values and Fare Outliers

import matplotlib.pyplot as plt
import seaborn as sns

# 1. Missing value count and percentage
missing_count = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df) * 100).round(2)
missing_summary = pd.DataFrame({"Missing_Count": missing_count, "Percent": missing_pct})

print("--- Missing Values Summary ---")
print(missing_summary[missing_summary["Missing_Count"] > 0])

# Plot missing counts bar chart
plt.figure(figsize=(6, 4))
sns.barplot(x=missing_count.index, y=missing_count.values, color="coral")
plt.title("Missing Values Count Per Column")
plt.xticks(rotation=45)
plt.savefig("missing_values.png", bbox_inches="tight")
plt.show()

# 2. Imputation: Fill missing Age with mean
print("\nAge missing before imputation:", df["Age"].isnull().sum())
df["Age"] = df["Age"].fillna(df["Age"].mean())
print("Age missing after mean imputation:", df["Age"].isnull().sum())

# 3. Fare Outlier Detection using IQR
q1 = df["Fare"].quantile(0.25)
q3 = df["Fare"].quantile(0.75)
iqr = q3 - q1

lower_limit = q1 - 1.5 * iqr
upper_limit = q3 + 1.5 * iqr

outliers = df[(df["Fare"] < lower_limit) | (df["Fare"] > upper_limit)]

print("\n Fare Outliers (1.5xIQR) ")
print("Q1:", round(q1, 2), "| Q3:", round(q3, 2), "| IQR:", round(iqr, 2))
print("Lower Limit:", round(lower_limit, 2), "| Upper Limit:", round(upper_limit, 2))
print("Outlier Count in Fare:", len(outliers))
```

task 8

```python

# FamilySize = SibSp (siblings/spouses) + Parch (parents/children) + 1 (the passenger themselves)
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

# IsAlone = 1 if passenger is travelling alone (FamilySize == 1), else 0
df["IsAlone"] = (df["FamilySize"] == 1).astype(int)

print(" Sample of New Features ")
print(df[["SibSp", "Parch", "FamilySize", "IsAlone"]].head())

pivot = df.pivot_table(index="Sex", columns="Pclass", values="Survived", aggfunc="mean")

print("\n Pivot Table: Survival Rate by Sex and Pclass ")
print(pivot.round(4))

unstacked = pivot.unstack()
print(f"\nHighest Survival Group : {unstacked.idxmax()} ({unstacked.max() * 100:.2f}%)")
print(f"Lowest Survival Group  : {unstacked.idxmin()} ({unstacked.min() * 100:.2f}%)")
```

```python
# Task 9: Visualizations and Observations

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# 1. Correlation Heatmap
plt.figure(figsize=(8, 6))
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.savefig("correlation_heatmap.png", bbox_inches="tight")
plt.show()

# 2. Survival Rate by Sex Plot & Print Rates
plt.figure(figsize=(6, 4))
sns.barplot(x="Sex", y="Survived", data=df, palette="pastel")
plt.title("Survival Rate by Sex")
plt.ylabel("Survival Rate")
plt.savefig("survival_by_sex.png", bbox_inches="tight")
plt.show()

male_surv = df[df["Sex"] == "male"]["Survived"].mean() * 100
female_surv = df[df["Sex"] == "female"]["Survived"].mean() * 100
print(f"Male Survival Rate  : {male_surv:.2f}%")
print(f"Female Survival Rate: {female_surv:.2f}%")

# 3. Scatter Plot: Age vs Fare (by Survival)
plt.figure(figsize=(8, 5))
sns.scatterplot(x="Age", y="Fare", hue="Survived", data=df, alpha=0.7, palette="Set1")
plt.title("Age vs Fare by Survival Status")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.savefig("age_vs_fare.png", bbox_inches="tight")
plt.show()

# 4. Save the cleaned dataset as required
df.to_csv("titanic_cleaned.csv", index=False)
print("Saved cleaned data to titanic_cleaned.csv!")

```
